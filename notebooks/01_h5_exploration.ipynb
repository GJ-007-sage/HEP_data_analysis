# Observable exploration

```python
def explore_observable(df, column_name, bins=50):

    values = df[column_name]

    print(f"\n=== {column_name} ===")

    print(f"Min   : {values.min()}")
    print(f"Max   : {values.max()}")
    print(f"Mean  : {values.mean()}")
    print(f"Std   : {values.std()}")

    plt.figure(figsize=(7,5))

    plt.hist(values, bins=bins)

    plt.xlabel(column_name)
    plt.ylabel("Events")

    plt.title(column_name)

    plt.show()
```

```python
# EXPLORE THE CENTRAL OBSERVABLES
```

```python
explore_observable(df, "Higgs_pT")
```

```python
explore_observable(df, "M_jj")
```

```python
explore_observable(df, "m_bbyy")
```

```python
explore_observable(df, "DPhi_bb")
```

```python
df["signed_DeltaPhi_jj_centered"] = np.where(
    df["signed_DeltaPhi_jj"] > np.pi,
    df["signed_DeltaPhi_jj"] - 2*np.pi,
    df["signed_DeltaPhi_jj"]
)

explore_observable(df, "signed_DeltaPhi_jj_centered")
```

```python
explore_observable(df, "costheta1")
```

```python
#helper function
def load_h5_to_df(filepath):

    with h5py.File(filepath, "r") as f:

        data = f["ForAnalysis/1d"][:]

    df = pd.DataFrame(data)

    return df
```

```python
sm_df = load_h5_to_df(sm_file)

chgtil_file = DATA_DIR / "new_input_bbyy_SMEFT_chgtil_4thMarch_2026.h5"

chgtil_df = load_h5_to_df(chgtil_file)
```

```python
plt.figure(figsize=(7,5))

plt.hist(
    sm_df["Higgs_pT"],
    bins=50,
    density=True,
    histtype="step",
    linewidth=2,
    label="SM"
)

plt.hist(
    chgtil_df["Higgs_pT"],
    bins=50,
    density=True,
    histtype="step",
    linewidth=2,
    label="chgtil"
)

plt.xlabel("Higgs_pT")
plt.ylabel("Normalized Events")

plt.legend()

plt.show()
```
